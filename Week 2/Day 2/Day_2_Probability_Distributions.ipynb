## Objective

Learn the basics of probability and probability distributions using NumPy and Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
coin_flips = np.random.choice([0, 1], size=10000)

heads = np.sum(coin_flips)
tails = len(coin_flips) - heads

print("Heads:", heads)
print("Tails:", tails)
print("Proportion of Heads:", heads / len(coin_flips))
```

## Coin Flip Simulation

A fair coin has an equal probability (0.5) of landing on heads or tails.

After simulating 10,000 coin flips, the proportion of heads was **0.5015**, which is very close to the expected probability of **0.5**.

This demonstrates the **Law of Large Numbers**, where the observed probability approaches the theoretical probability as the number of trials increases.

```python
data = np.random.normal(loc=0, scale=1, size=10000)

plt.figure(figsize=(8,5))
plt.hist(data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Normal Distribution

The histogram shows a bell-shaped distribution, where most values are concentrated around the mean. As values move farther from the center, they become less frequent. This is the characteristic shape of a normal (Gaussian) distribution.

```python
python_students = 60
ai_students = 40
both = 30

conditional_probability = both / ai_students

print("P(Python | AI) =", conditional_probability)
```

```python
students = np.arange(100)

python_group = set(np.concatenate((np.arange(0, 30), np.arange(30, 60))))
ai_group = set(np.concatenate((np.arange(30, 60), np.arange(60, 70))))

intersection = python_group & ai_group

simulation = len(intersection) / len(ai_group)

print("Simulated Conditional Probability:", simulation)
```

## Conditional Probability

The conditional probability **P(Python | AI)** is **0.75**.

This means that among students who study AI, **75% also study Python**. The simulation produced the same result as the manual calculation, confirming the formula.

```python
P_A = 0.6
P_B_given_A = 0.5
P_B = 0.4

P_A_given_B = (P_B_given_A * P_A) / P_B

print("Bayes Result:", P_A_given_B)
```

## Bayes' Theorem

Bayes' Theorem updates the probability of an event after observing new evidence.

In this example, the posterior probability is **0.75**, meaning the evidence increases our confidence in the event.
