## Objective

Learn how vectors, matrices, dot products, and matrix multiplication are used in machine learning.

```python
import numpy as np
```

```python
customer = np.array([25, 50000, 3])

print(customer)
print("Shape:", customer.shape)
```

```python
X = np.array([
    [25, 50000, 3],
    [40, 80000, 10],
    [33, 62000, 5]
])

print(X)
print("Shape:", X.shape)
```

## Vectors and Matrices

A vector represents one data sample with its features.

A matrix represents multiple data samples, where each row is a sample and each column is a feature.

```python
weights = np.array([0.1, 0.0002, 1.5])

print(weights)
```

```python
prediction = np.dot(customer, weights)

print("Prediction:", prediction)
```

## Dot Product

The dot product multiplies each feature by its corresponding weight and sums the results.

For this customer:

- Age contribution = 25 × 0.1 = 2.5
- Income contribution = 50000 × 0.0002 = 10
- Tenure contribution = 3 × 1.5 = 4.5

Final prediction = **17.0**

This operation is the foundation of many machine learning models such as Linear Regression and Logistic Regression.

```python
predictions = X @ weights

print("Predictions:")
print(predictions)
```

## Matrix Multiplication

Matrix multiplication allows us to compute predictions for multiple samples at once.

Instead of calculating the dot product for each sample separately, we multiply the entire dataset by the weight vector, producing one prediction for each row.

```python
wrong_weights = np.array([0.1, 0.0002])

X @ wrong_weights
```

## Shape Mismatch

The error occurs because the matrix and the weight vector have incompatible shapes.

- Matrix shape: (3, 3)
- Weight vector shape: (2,)

Matrix multiplication requires the number of columns in the matrix to match the number of elements in the weight vector.

To fix the error, the weight vector must contain three values, one for each feature.
