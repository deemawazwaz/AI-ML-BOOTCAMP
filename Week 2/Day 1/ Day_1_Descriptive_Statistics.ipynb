# Week 2 - Day 1: Descriptive Statistics

## Objective
Compute and interpret descriptive statistics including:
- Mean
- Median
- Mode
- Variance
- Standard Deviation
- Interquartile Range (IQR)

Dataset: tested.csv

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("tested.csv")
df.head()
```

```python
df.columns
```

```python
age = df["Age"].dropna()

print("Mean:", age.mean())
print("Median:", age.median())
print("Mode:", age.mode()[0])
print("Variance:", age.var())
print("Standard Deviation:", age.std())

q1 = age.quantile(0.25)
q3 = age.quantile(0.75)
iqr = q3 - q1

print("IQR:", iqr)
```

```python
print("Minimum Age:", age.min())
print("Maximum Age:", age.max())
```

## Which Measure Better Represents the Typical Age?

The **median** is a better measure of the typical age because the Age column contains a wide range of values (from 0.17 to 76 years). These extreme values can influence the mean, while the median is more robust and better represents the center of the data.

## Summary

The **Age** column has:

- Mean: **30.27**
- Median: **27.00**
- Mode: **21.00**
- Standard Deviation: **14.18**
- IQR: **18.00**

The ages are spread across a wide range, from **0.17** to **76** years. The median provides a better representation of a typical passenger's age because it is less affected by extreme values than the mean.
